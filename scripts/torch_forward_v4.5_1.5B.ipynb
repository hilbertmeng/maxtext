## chunk decode 生成

```python

DEVICE = 'cpu'
import math
def pad_inputs(input_ids, prefill_chunk_size):
    batch_size, seq_length = input_ids.shape
    prefill_length = input_ids.shape[1]
    chunks = math.ceil(prefill_length / prefill_chunk_size)
    prefill_pad_length = prefill_chunk_size * chunks - prefill_length
    input_ids = torch.nn.functional.pad(input_ids, pad=[0, prefill_pad_length]).to(DEVICE)
    input_pos = torch.arange(seq_length + prefill_pad_length, device=DEVICE).unsqueeze(0)
    gen_index = seq_length - 1
    return input_ids, input_pos, gen_index, chunks

    
loss_fn = torch.nn.CrossEntropyLoss(reduction='none')

batch_id = [0]
k = 0
prefill_chunk_size = 256
input_ids = torch.tensor(input_ids)
x, input_pos, gen_index, chunks = pad_inputs(input_ids[:, :-1], prefill_chunk_size)
labels, _, _, _ = pad_inputs(input_ids[:, 1:], prefill_chunk_size)
g = (gen_index + k) % prefill_chunk_size
losses = []
print(f'x: {x.shape} labels: {labels.shape}')
for j in range(chunks):
    kvshift_cache_pos = [0 if j == 0 else 1]
    kvshift_cache_pos.append(g if j == chunks - 1 else -1)
    last_chunk = True if j == chunks - 1 else False
    batch_indexes = torch.tensor([0])
    kvshift_cache_pos = torch.tensor(kvshift_cache_pos).unsqueeze(0)
    chunk_input_ids = x[:, j*int(prefill_chunk_size): (j+1) * int(prefill_chunk_size)]
    chunk_labels = labels[:, j*int(prefill_chunk_size): (j+1) * int(prefill_chunk_size)]
    chunk_de_mudd_embeds = [de_emb(chunk_input_ids) for de_emb in model.de_embeddings]
    print(f'chunk_de_mudd_embeds: {len(chunk_de_mudd_embeds)} tensors, each shape: {chunk_de_mudd_embeds[0].shape}')
    
    with torch.no_grad():
        output = model(chunk_input_ids, 
                       input_pos[:, j*int(prefill_chunk_size): (j+1) * int(prefill_chunk_size)], 
                         batch_indexes=batch_indexes, 
                         kvshift_cache_poss=kvshift_cache_pos,
                         de_mudd_embeds=chunk_de_mudd_embeds,
                         labels=chunk_labels, 
                       last_chunk=last_chunk
                            )
    logits = output.logits
    loss = loss_fn(output.logits.view(-1, output.logits.shape[-1]), chunk_labels.view(-1))
    loss_mean = round(loss.mean().item(), 4)
    losses.append(loss_mean)
    print(f'{j}-loss: {loss_mean}')
```
